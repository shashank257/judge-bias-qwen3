# Week 1 Checklist — Data, Validation, and Hosted API Selection

**Target:** Finish data preparation and validation in Google Colab, prepare experiment inputs, and select a hosted API for Qwen. By the end of the week, the datasets should be ready for inference.

## 1. Colab workspace and dataset records

- [ ] Create Colab notebooks for data preparation, validation, and API compatibility checks.
- [ ] Set up persistent storage for original datasets, processed datasets, validation reports, and API responses.
- [ ] Record dataset sources, licenses, versions/revisions, and download dates.
- [ ] Use seed **261** for sampling and maintain a preprocessing version log.
- [ ] Define a common record format: prompt ID, pair ID, source, question, original answers, human preference, answer lengths, and markdown indicators.

## 2. Main Arena dataset

- [ ] Load the **2023 Chatbot Arena human-preference dataset** and inspect its fields and vote labels.
- [ ] Implement filters: English, single-turn conversations, non-tied human preferences, and answer-length ratio no greater than **2**.
- [ ] Define how language, single-turn eligibility, and answer length are measured.
- [ ] Identify missing answers, invalid labels, duplicate records, and repeated prompts; preserve prompt IDs for later clustering.
- [ ] Exclude answers containing tables or code blocks according to the proposal's content-preservation rule.
- [ ] Report retained and excluded record counts after every filtering step.
- [ ] Create separate position-bias and format-bias pools; require markdown in **both** original answers for the format pool.
- [ ] Verify at least **1,000 eligible pairs per bias**. If needed, apply the same filters to the proposed 2025 Arena supplement and retain source labels.
- [ ] Freeze selected pairs so every model configuration receives the same relevant examples.

## 3. Presentation changes and validation

- [ ] Define supported markdown features: headings, bold text, and bullet markers.
- [ ] Implement deterministic markdown stripping without rewriting answer content.
- [ ] Define a word-sequence comparison that permits removal of supported formatting markers but detects added, removed, or reordered words.
- [ ] Validate every stripped answer automatically; save failures and resolve or exclude affected pairs before freezing the pool.
- [ ] Manually audit **50 pairs** and record whether meaning, wording, and list structure remain interpretable.
- [ ] Generate all four format combinations: markdown/markdown, markdown/plain, plain/markdown, and plain/plain.
- [ ] Generate both answer orders for each combination.
- [ ] Preserve original answer identities when swapping order so verdicts map back correctly.

## 4. Remaining evaluation data and controls

- [ ] Load and validate the proposed **MT-Bench expert-vote data**, checking question, answer, turn, and label alignment.
- [ ] Prepare human-agreement labels for later accuracy and Cohen's kappa calculations.
- [ ] Implement random-verdict and length-only baselines.
- [ ] Measure whether human preferences correlate with markdown presence.
- [ ] Load UltraFeedback at a pinned revision and select **500 eligible prompts** from ShareGPT, UltraChat, and Evol-Instruct sources, excluding TruthfulQA-derived prompts.
- [ ] Verify that each selected UltraFeedback prompt has four usable answers and the required fine-grained ratings.
- [ ] Calculate gold-proxy quality as the **mean of the four fine-grained ratings**; document missing-rating and tie handling.
- [ ] Label markdown presence and answer-generating model; prepare the sensitivity subset excluding prompts containing GPT-4 answers.
- [ ] Save reproducible candidate subsets and bracket definitions for best-of-n with **n = 1, 2, 4**.

## 5. Experiment specification

- [ ] Finalize the judge prompt, answer delimiters, verdict format, and rules for ties or invalid responses.
- [ ] Create a request manifest identifying each pair, presentation condition, order, model configuration, and repeat.
- [ ] Include independent repeated judgments of identical inputs in each order for noise-floor measurement.
- [ ] Ensure intentional repeats trigger fresh generations rather than reuse saved answers.
- [ ] Calculate request counts from the manifest, accounting for comparisons reused across experiments.
- [ ] Finalize definitions of excess flip rate, human agreement, swap-and-average agreement, format drift, and selection regret.
- [ ] Record later analysis requirements: prompt-clustered bootstrap intervals, size/thinking interaction analysis, Holm correction, and exploratory correlations.
- [ ] Define how failures, retries, missing verdicts, and truncated reasoning will be recorded and analyzed.

## 6. Hosted API comparison

- [ ] Compare **Together AI, OpenRouter, and other relevant hosted providers** using their current catalogs.
- [ ] Check the exact original hybrid **Qwen3 0.6B, 1.7B, 4B, and 8B** checkpoints and record exact API model identifiers.
- [ ] Verify thinking **on and off with the same weights** for each checkpoint.
- [ ] Check sampling controls, context limits, output limits, reasoning-token visibility, seed support, and the proposed **2,048-token thinking budget**.
- [ ] Check whether the proposal's truncation-and-verdict procedure is possible through the API; document required changes.
- [ ] Record serving provider, model revision, and precision/quantization where disclosed; check whether they can remain fixed.
- [ ] For OpenRouter, verify provider pinning, disable unintended fallbacks, and require support for requested parameters.
- [ ] Verify that thinking off disables reasoning rather than merely hides reasoning text.
- [ ] Compare input/output pricing, reasoning-token billing, rate limits, availability, and expected total experiment cost.

**Reference documentation:** [Together AI model catalog](https://docs.together.ai/docs/serverless/models) · [OpenRouter provider routing](https://openrouter.ai/docs/guides/routing/provider-selection) · [OpenRouter reasoning controls](https://openrouter.ai/docs/guides/best-practices/reasoning-tokens)

## 7. API validation and final deliverables

- [ ] Make small compatibility calls from Colab to verify checkpoints, thinking controls, verdict parsing, and response metadata.
- [ ] Complete the proposed **100-pair pilot on two sizes**, covering both thinking modes; measure latency, token usage, failures, truncation, and cost.
- [ ] Select the API based on experiment compatibility and pilot results.
- [ ] Document unavailable checkpoints or unsupported controls before finalizing the model lineup; do not silently substitute models.
- [ ] Update the proposal for **Colab plus hosted APIs**, replacing Kaggle/vLLM/GPU-hour assumptions with API cost and throughput estimates.
- [ ] Freeze the evaluation contract after the pilot: datasets, model IDs, provider settings, prompts, sampling parameters, metrics, and analysis rules.
- [ ] Save final data exports, validation report, manual audit, request manifest, API comparison, pilot results, and updated proposal.

**Design constraint:** Exact model availability must be established early. Identical serving backends and FP16 precision cannot be assumed with hosted APIs; record any differences or unavailable information in the revised experimental design.
