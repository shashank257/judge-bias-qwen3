# UltraFeedback · final six-mode inference
500 frozen prompts and four existing answers per prompt. The 25-prompt pilot contributes its recorded judgments. Run setup in Colab with a GPU and Google Drive, then choose the model and thinking-mode parts to execute. Run All skips finished parts. Each batch is committed to Drive. The original eight-mode contract is retained for compatibility with the project's prior checkpoints. Executable parts and the final analysis cover 0.6B, 1.7B and 4B only.

To replay the published results on CPU, open REPRODUCE_RESULTS.ipynb at the package root. This GPU notebook samples new judgments for work that has no saved checkpoint. Independent generations can differ from the published observations.

```python
from pathlib import Path
import subprocess

# A local checkout works offline. A fresh Colab runtime checks out the final study from GitHub.
cwd = Path.cwd().resolve()
study_path = Path('ultrafeedback/final')
candidates = [p for parent in (cwd, *cwd.parents) for p in (parent, parent / study_path)]
root = next((p for p in candidates if (p / 'results/latest_decisions.csv').is_file()
             and (p / 'reproduce.py').is_file()), None)
if root is None:
    checkout = cwd / 'judge-bias-qwen3'
    if not (checkout / '.git').is_dir():
        subprocess.run(['git', 'clone', '--depth=1', '--filter=blob:none', '--sparse',
                        'https://github.com/shashank257/judge-bias-qwen3.git', str(checkout)], check=True)
    subprocess.run(['git', '-C', str(checkout), 'sparse-checkout', 'set', '--no-cone',
                    '/ultrafeedback/final/*', '!/ultrafeedback/final/UltraFeedback_Final.zip'], check=True)
    root = checkout / study_path
print('Using', root)

from google.colab import drive
import hashlib
import sys
import tempfile
import zipfile

drive.mount('/content/drive')
OUT = Path('/content/drive/MyDrive/DSC261_UltraFeedback/full_500_recovery_v1')
PROJECT = Path(tempfile.mkdtemp(prefix='uf-final-inputs-'))
frozen = root / 'inference/uf_frozen_500_b95fe3a5d9b7.bin'
payload = frozen.read_bytes()
if hashlib.sha256(payload).hexdigest() != 'b95fe3a5d9b7160ef814e281a0cb5fe5afdc8ef85605ed98c4aa372ab680e75f':
    raise ValueError('Frozen study input checksum mismatch')
with zipfile.ZipFile(frozen) as archive:
    for name in ('full.jsonl', 'pilot.jsonl', 'configs.json', 'source_trial_manifest.json',
                 'recovery_metadata.json', 'recovery_audit.jsonl'):
        archive.extract(name, PROJECT)

# Import the editable Python modules from this checkout.
sys.path.insert(0, str(root / 'code'))
for name in list(sys.modules):
    if name == 'src' or name.startswith('src.'):
        del sys.modules[name]
from src.uf_full import FullRun, atomic

run = FullRun(PROJECT, OUT)
backup = OUT / 'frozen_inputs_and_pilot_eight_modes.zip'
if not backup.exists():
    atomic(backup, payload)
print(f'Ready: {len(run.prompts)} prompts. Saving checkpoints in {OUT}.')
```

```python
for c in (cfg for cfg in run.configs if cfg['checkpoint'] != 'Qwen/Qwen3-8B'):
    print(c['config_id'])
    for i in range(1,len(run.parts(c))+1):
        folder=OUT/c['config_id']/f'part-{i:02}'
        state='finished' if (folder/'finished.json').exists() else 'resumable' if folder.exists() else 'ready'
        print(f'{i:02} {state}',end=' | ')
    print()
```
